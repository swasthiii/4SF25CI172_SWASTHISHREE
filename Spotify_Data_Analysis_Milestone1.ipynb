# Spotify Data Analysis – Milestone 1

**Dataset notice:** This is synthetic Spotify-style practice data created for learning. It is not an official Spotify export or real Spotify catalogue data.

**Tools:** Python, Pandas, NumPy, Matplotlib, Seaborn. Keep `spotify_data clean.csv` in this folder and run cells from top to bottom.

## 1. Import required libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
pd.set_option("display.max_columns", None)
print("Libraries imported successfully!")
```

## 2. Read the CSV file

```python
file_name = "spotify_data clean.csv"
spotify = pd.read_csv(file_name)

print("Dataset loaded successfully!")
print("Rows and columns:", spotify.shape)
display(spotify.head())
```

## 3. Understand the dataset

```python
spotify.info()
display(spotify.describe(include="all").T)
print("Missing values:")
display(spotify.isnull().sum().to_frame("Missing Values"))
print("Duplicate rows:", spotify.duplicated().sum())
```

## 4. Clean and transform the data

```python
df = spotify.copy()
df = df.drop_duplicates()
df.columns = df.columns.str.strip()

if "album_release_date" in df.columns:
    df["album_release_date"] = pd.to_datetime(df["album_release_date"], errors="coerce")
    df["release_year"] = df["album_release_date"].dt.year

if "duration_ms" in df.columns:
    df["duration_minutes"] = df["duration_ms"] / 60000

display(df.head())
print("Cleaned dataset shape:", df.shape)
```

## 5. Handle missing values

```python
print("Missing values before handling:")
display(df.isnull().sum().to_frame("Missing Before"))

numeric_columns = df.select_dtypes(include=np.number).columns
for column in numeric_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna(df[column].mean())

text_columns = df.select_dtypes(include=["object", "string"]).columns
for column in text_columns:
    if df[column].isnull().any():
        df[column] = df[column].fillna("Unknown")

print("Missing values after handling:")
display(df.isnull().sum().to_frame("Missing After"))
```

## 6. Generate insights

```python
if "track_popularity" in df.columns:
    print("Average track popularity:", round(df["track_popularity"].mean(), 2))

if "artist_name" in df.columns:
    print("\nTop 10 artists by number of tracks:")
    display(df["artist_name"].value_counts().head(10).to_frame("Number of Tracks"))

if {"album_type", "track_popularity"}.issubset(df.columns):
    print("\nAverage popularity by album type:")
    display(df.groupby("album_type")["track_popularity"].mean().sort_values(ascending=False).round(2).to_frame("Average Popularity"))

if {"explicit", "track_popularity"}.issubset(df.columns):
    print("\nAverage popularity by explicit status:")
    display(df.groupby("explicit")["track_popularity"].mean().round(2).to_frame("Average Popularity"))

if "release_year" in df.columns:
    print("\nTracks by release year:")
    display(df["release_year"].value_counts().head(10).to_frame("Number of Tracks"))
```

## 7. Visualizations

### 7.1 Top 10 artists by number of tracks

```python
top_artists = df["artist_name"].value_counts().head(10)
plt.figure(figsize=(10,5))
sns.barplot(x=top_artists.values, y=top_artists.index)
plt.title("Top 10 Artists by Number of Tracks")
plt.xlabel("Number of Tracks")
plt.ylabel("Artist")
plt.tight_layout()
plt.show()
```

### 7.2 Track popularity distribution

```python
plt.figure(figsize=(8,5))
sns.histplot(df["track_popularity"].dropna(), bins=20, kde=True)
plt.title("Distribution of Track Popularity")
plt.xlabel("Track Popularity")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.show()
```

### 7.3 Explicit and non-explicit tracks

```python
counts = df["explicit"].value_counts()
plt.figure(figsize=(6,6))
plt.pie(counts.values, labels=counts.index.astype(str), autopct="%1.1f%%", startangle=90)
plt.title("Explicit vs Non-Explicit Tracks")
plt.tight_layout()
plt.show()
```

### 7.4 Tracks released each year

```python
yearly = df["release_year"].dropna().astype(int).value_counts().sort_index()
plt.figure(figsize=(12,5))
plt.plot(yearly.index, yearly.values, marker="o")
plt.title("Number of Tracks Released Each Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Tracks")
plt.tight_layout()
plt.show()
```

## 8. Final summary

```python
print("FINAL DATASET SUMMARY")
print("---------------------")
print("Total rows:", df.shape[0])
print("Total columns:", df.shape[1])
print("Duplicate rows remaining:", df.duplicated().sum())
print("Missing values remaining:", int(df.isnull().sum().sum()))
print("Average track popularity:", round(df["track_popularity"].mean(), 2))
print("Artist with most tracks:", df["artist_name"].value_counts().idxmax())
```
